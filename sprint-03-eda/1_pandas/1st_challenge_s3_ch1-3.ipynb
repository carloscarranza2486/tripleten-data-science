# 🎯 Reto: Análisis de una Tienda de Videojuegos

Imagina que tienes un dataset de ventas de videojuegos con algunos problemas de calidad de datos. Tu misión es limpiarlo y analizarlo.

### Paso 1: Preparación y Limpieza Básica
Crea un DataFrame con estos datos (copia y pega este código):

```python
import pandas as pd
import numpy as np

# Datos con problemas intencionalmente
data = {
    'game_name': ['FIFA 2023', 'Call of Duty', 'FIFA 2023', 'Minecraft', 'Call of Duty', 'Fortnite', None, 'Minecraft'],
    'genre': ['Sports', 'Action', 'Sports', 'Sandbox', 'Action', 'Battle Royale', 'RPG', 'Sandbox'],
    'price': [59.99, 69.99, 59.99, 26.95, 69.99, 0.00, 49.99, None],
    'sales_count': [1500, 2300, 1500, 800, 2300, 3200, 150, 800]
}

games_df = pd.DataFrame(data)
```

Tu tarea:
- Renombra la columna 'sales_count' a 'ventas'
- Identifica cuántos valores ausentes hay en total
- Rellena los valores ausentes de 'price' con el precio promedio

```python
games_df.rename(columns = {'sales_count':'ventas'}, inplace=True)
print(games_df.columns)
games_df.fillna(games_df['price'].mean(), inplace=True)
print(games_df.isna().sum())
```

### Paso 2: Manejo de Duplicados
- Identifica cuántas filas duplicadas completas existen
- Elimina los duplicados manteniendo la primera ocurrencia
- Verifica cuántas filas te quedaron después de la limpieza

```python
print(games_df.duplicated().sum())
games_df.drop_duplicates(inplace=True)
print(games_df.duplicated().sum())
```

### Paso 3: Análisis de Valores Únicos
- Encuentra cuántos géneros únicos hay
- Muestra la lista de todos los juegos únicos
- Identifica cuál es el precio más alto y más bajo

```python
print(games_df['genre'].nunique())
print(games_df['genre'].unique())
print(games_df['price'].max())
print(games_df['price'].min())
```

### Paso 4: Agrupación y Análisis
- Agrupa por 'genre' y calcula el total de ventas por género
- Ordena los resultados de mayor a menor ventas
- Identifica qué género tiene las mayores ventas totales

```python
print(games_df.groupby('genre')['ventas'].sum())
print(games_df.sort_values(by='ventas', ascending=False))
print(games_df.groupby('genre')['ventas'].sum().idxmax())
```
