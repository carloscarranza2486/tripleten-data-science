# Filtered by multiple conditions

## Guided practice

#### Exercise 1: Filtering with boolean mask

```python
import pandas as pd

df = pd.read_csv('/datasets/vg_sales.csv')
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')

df_filtered = df[(df['year_of_release'] >= 1980) & (df['year_of_release'] < 1990)]
print(df_filtered.head())
```

#### Exercise 2: Filtering with query and multiple conditions

```python
import pandas as pd

df = pd.read_csv('/datasets/vg_sales.csv')
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')

q_string = "na_sales >= 1 or eu_sales >= 1 or jp_sales >= 1"
print(df.query(q_string).head())
```

#### Exercise 1: filtering with a consult string assigned to a query

```python
import pandas as pd

df = pd.read_csv('/datasets/vg_sales.csv')
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')

# Selecciono las columnas solicitadas
cols = ['name', 'developer', 'na_sales', 'eu_sales', 'jp_sales']

# Filtro por valores mayores a cero
q_string = "na_sales > 0 and eu_sales > 0 and jp_sales > 0"

# Filtro por las columnas y asigno el resultado a df_filtered
df_filtered = df.query(q_string)[cols]

# Imprimo el data frame completo
print(df_filtered)
```

#### Exercise 2: filtering values if sales in JP are higher than eu+na 

```python
import pandas as pd

df = pd.read_csv('/datasets/vg_sales.csv')
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')

developers = ['SquareSoft', 'Enix Corporation', 'Square Enix']
cols = ['name', 'developer', 'na_sales', 'eu_sales', 'jp_sales']

# Incopora los cambios necesario para reflejar las nuevas restricciones
q_string = " jp_sales > 0 and na_sales > 0 and eu_sales > 0 and (jp_sales > (na_sales + eu_sales)) and developer in @developers " 

df_filtered = df.query(q_string)[cols]
print(df_filtered ) 
```
