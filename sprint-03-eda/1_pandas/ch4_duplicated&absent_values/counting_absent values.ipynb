## Practice activity

#### Exercise 1: inversing filtering with ~

```python
import pandas as pd

df_logs = pd.read_csv('/datasets/visit_log.csv')

df_logs['email'].isna()
~df_logs['email'].isna()
df_emails = df_logs[~df_logs['email'].isna()]
print(df_emails.head(10))
```

#### Exercise 2: data validation

```python
import pandas as pd

df_logs = pd.read_csv('/datasets/visit_log.csv')

df_emails = df_logs[df_logs['email'].isna() & df_logs['source'].isna()]

print(df_emails)
```
