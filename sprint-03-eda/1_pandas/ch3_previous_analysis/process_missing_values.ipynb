# Process missing values

#### Exercise 1: absent values with "isna" & "sum" method

```python
import pandas as pd

df = pd.read_csv('/datasets/music_log_raw.csv')

mis_val = df.isna().sum()

print(mis_val)
```

#### Exercise2: replace values if they are empty

```python
import pandas as pd

df = pd.read_csv('/datasets/music_log_raw.csv')

columns_to_replace = ['genre', 'Artist', 'track']

for col in columns_to_replace:
	df[col] = df[col].fillna('no_info')

print(df.isna().sum())
```

#### Exercise 3: replace empty spaces in a no-string column

```python
import pandas as pd

df = pd.read_csv('/datasets/music_log_raw.csv')

df['total play'].fillna(0, inplace = True)


print(df.isna().sum())
```

#### Exercise 4: drop rows

```python
import pandas as pd

df = pd.read_csv('/datasets/music_log_raw.csv')

df.dropna(subset=['total play'], inplace = True)

print(df.isna().sum())
```
