# Frequency histograms

### Example: 2 histograms using different methodologies

```python
import pandas as pd

data = pd.Series([11, 20, 22, 31, 32, 33, 41, 42, 43, 44, 51, 52, 53, 54, 55, 61, 62, 63, 64, 65, 66, 71, 72, 73, 74, 75, 76, 77, 81, 82, 83, 84, 85, 86, 87, 88, 91, 92, 93, 94, 95, 96, 97, 98, 99])

data.hist(bins=4, alpha=0.5)  # crea un histograma con cuatro contenedores

data.hist(
    bins=[11, 20, 30, 40, 50, 60, 70, 80, 90, 99], alpha=0.7
)  # construye un histograma con nueve contenedores cuyos límites estén enumerados, donde el argumento alfa nos da un gráfico opaco
```

Al pasar una lista como valor para el parámetro bins=, creamos nueve contenedores: de 11 a 20, de 20 a 30, y así sucesivamente. El parámetro alpha= se usa para ajustar la transparencia, lo que permite que los dos gráficos se muestren en uno solo.

# Density histograms

La curva de densidad mostrada conecta los picos de las barras e ilustra la distribución de los tiempos de retraso de llegada en minutos.  Las barras se mantienen junto a la curva para mostrar la relación entre la curva de densidad y el histograma de densidad, aunque ya no son necesarias.

El funcionamiento es el mismo que en los histogramas de densidad: el área bajo la curva entre dos valores representa la frecuencia de los valores en ese intervalo. Sin embargo, hay una diferencia en el eje Y. ¿Te has fijado? Ahora los valores están en un intervalo diferente, y el título del eje es “Density”. ¿Qué cambió? No es magia, sino otra convención común para visualizar la densidad de frecuencia.

Los números del eje Y ahora representan porcentajes. Por ejemplo, 0.01 equivale al 1% del área total bajo la curva, 0.02 al 2%, y así sucesivamente. El área total representa el número total de casos en nuestro dataset.

En resumen, tanto los histogramas de densidad como las curvas de densidad transmiten la misma información que los histogramas de frecuencia normales: las frecuencias. La diferencia clave es que los valores de frecuencia reales ahora corresponden a las áreas.

```python
import numpy as np

x = [1, 2, 3, 4, 5, 6] # dataset 1
y = [41, 62, 89, 96, 108, 115] # dataset 2

covariance_matrix = np.cov(x,y) # calculamos la matriz de covarianza
covariance = covariance_matrix[0][1] # extraemos la covarianza como valor
print(covariance)
```

# Standard deviation and normal distribution

```python
import numpy as np

x = [1, 2, 3, 4, 5, 6] # dataset

standard_deviation = np.std(x)
print(standard_deviation)
```

```python
import numpy as np

variance = 3.5
standard_deviation = np.sqrt(variance)
print(standard_deviation)
```

*La principal diferencia es que la desviación estándar es la raíz cuadrada de la varianza. La varianza mide el promedio de las desviaciones al cuadrado respecto a la media, usando unidades cuadradas, mientras que la desviación estándar indica la dispersión en las mismas unidades originales de los datos, siendo más interpretable.*
