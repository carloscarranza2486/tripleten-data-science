### Exercise 1: Probability - Shuffle play - Spotify

```python
import pandas as pd

cool_rock = pd.DataFrame(
    {
        'Artist': [
            'Queen',
            'Queen',
            'Queen',
            'Pink Floyd',
            'Nirvana',
            'AC/DC',
            'AC/DC',
            'Scorpions',
            'Scorpions',
            'Scorpions',
        ],
        'Song': [
            'The Show Must Go On',
            'Another One Bites The Dust',
            'We Will Rock You',
            'Wish You Were Here',
            'Smells Like Teen Spirit',
            'Highway To Hell',
            'Back in Black',
            'Wind Of Change',
            'Still Loving You',
            'Send Me An Angel',
        ],
    }
)

desired_artist = 'Queen'

desired_outcomes = len(cool_rock[cool_rock["Artist"]=="Queen"])
total_outcomes = len(cool_rock)
probability = desired_outcomes / total_outcomes
print(probability)
```

## Factorial

```python
from math import factorial

candidates_amount = 5

lists_amount = factorial(candidates_amount)
print(lists_amount)
```

```python
from math import factorial

n = 5
k = 3

combinations = factorial(n) / (factorial(k)*(factorial(n-k)))
print(combinations) # Output: 10.0
```

### Exercises

#### Exercise 1: Calculate probability to earn a prize

```python
from math import factorial

tasks = 10
permutations = factorial(tasks)
probability = 1/permutations

print(probability)
```

### Exercise 2: Guess the "secret combination"

```python
from math import factorial

tasks = 10
chosen = 3

combinations = factorial(tasks) / (factorial(chosen) * factorial(tasks-chosen))
probability = 1 / combinations

print(probability)
```
