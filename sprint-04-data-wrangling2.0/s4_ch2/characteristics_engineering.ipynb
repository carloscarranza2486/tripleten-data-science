#### Exercise 3: Group by new column and calculate sales using sum()

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/datasets/vg_sales.csv')

def score_group(score):
    if score < 60:
        return 'low'
    elif score < 80:
        return 'medium'
    elif score >= 80:
        return 'high'
    else:
        return 'no score'

df['score_categorized'] = df['critic_score'].apply(score_group)

df_grouped = df.groupby('score_categorized')
df_sum = df_grouped['na_sales'].sum()

print(df_sum)
```
