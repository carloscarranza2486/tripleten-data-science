#### Exercise 1: Using format parameter

```python
import pandas as pd

string_date = '5/13/13 12:04:00'
fecha = pd.to_datetime(string_date, format='%m/%d/%y %H:%M:%S')
print(fecha)
```

#### Exercise 2

```python
import pandas as pd

raw_date = '20-12-2002Z04:31:00'
clean_date = pd.to_datetime(raw_date, format='%d-%m-%YZ%H:%M:%S')
print(clean_date)
```
