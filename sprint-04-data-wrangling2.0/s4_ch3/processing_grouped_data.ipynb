# Processing grouped data with agg()

#### Exercise 1: Using agg() method 

```python
import pandas as pd

df = pd.read_csv('/datasets/vg_sales.csv')
df['total_sales'] = df['na_sales'] + df['eu_sales'] + df['jp_sales']

grp = df.groupby('genre')

agg_dict = {'total_sales': 'sum','na_sales': 'mean','eu_sales': 'mean', 'jp_sales': 'mean'}
genre = grp.agg(agg_dict)

print(genre)
```

## Pivot Tables

Exercise 1: Create a pivot table to filter by games launched after 2000s

```python
import pandas as pd

df = pd.read_csv('/datasets/vg_sales.csv')
df = df[df['year_of_release'] >= 2000]

df_pivot = df.pivot_table(index='genre',
                         columns= 'year_of_release',
                         values= 'jp_sales',
                         aggfunc= 'mean')

print(df_pivot)
```
